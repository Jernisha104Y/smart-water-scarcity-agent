```python
import pandas as pd

all_records = []

for page_number in tn_pages:
    text = pdf[page_number].get_text()
    lines = [line.strip() for line in text.splitlines() if line.strip()]

    # Find the first actual Tamil Nadu record
    try:
        start = lines.index("Tamil Nadu")
    except ValueError:
        continue

    data_lines = lines[start:]

    # Records contain 8 fields
    for i in range(0, len(data_lines) - 7, 8):
        record = data_lines[i:i+8]

        if len(record) == 8 and record[0] == "Tamil Nadu":
            all_records.append(record)

columns = [
    "State",
    "District",
    "Block",
    "Village",
    "Latitude",
    "Longitude",
    "Date",
    "WL_mbgl"
]

groundwater_df = pd.DataFrame(all_records, columns=columns)

# Convert numeric fields
groundwater_df["Latitude"] = pd.to_numeric(
    groundwater_df["Latitude"], errors="coerce"
)

groundwater_df["Longitude"] = pd.to_numeric(
    groundwater_df["Longitude"], errors="coerce"
)

groundwater_df["WL_mbgl"] = pd.to_numeric(
    groundwater_df["WL_mbgl"], errors="coerce"
)

# Convert date
groundwater_df["Date"] = pd.to_datetime(
    groundwater_df["Date"],
    format="%d-%m-%y",
    errors="coerce"
)

print("Total records:", len(groundwater_df))
print("\nColumns:")
print(groundwater_df.columns.tolist())

print("\nFirst 5 records:")
print(groundwater_df.head())
```

```python
print("=== DATA QUALITY CHECK ===")

print("\nShape:")
print(groundwater_df.shape)

print("\nMissing values:")
print(groundwater_df.isna().sum())

print("\nDate range:")
print(groundwater_df["Date"].min(), "to", groundwater_df["Date"].max())

print("\nNumber of districts:")
print(groundwater_df["District"].nunique())

print("\nNumber of villages:")
print(groundwater_df["Village"].nunique())

print("\nDuplicate rows:")
print(groundwater_df.duplicated().sum())

print("\nLatitude range:")
print(groundwater_df["Latitude"].min(), "to", groundwater_df["Latitude"].max())

print("\nLongitude range:")
print(groundwater_df["Longitude"].min(), "to", groundwater_df["Longitude"].max())

print("\nGroundwater level statistics (mbgl):")
print(groundwater_df["WL_mbgl"].describe())
```

```python
missing_rows = groundwater_df[
    groundwater_df["Latitude"].isna()
    | groundwater_df["Longitude"].isna()
    | groundwater_df["Date"].isna()
    | groundwater_df["WL_mbgl"].isna()
]

print("Rows with missing important values:", len(missing_rows))

print("\nMissing rows:")
print(missing_rows.to_string(index=False))
```

```python
# Check exact duplicate rows
exact_duplicates = groundwater_df.duplicated(keep=False)

print("Total rows:", len(groundwater_df))
print("Exact duplicate rows:", exact_duplicates.sum())

print("\nUnique rows:", groundwater_df.drop_duplicates().shape[0])

print("\nRows with missing Date/WL:")
print(
    groundwater_df[
        groundwater_df["Date"].isna() | groundwater_df["WL_mbgl"].isna()
    ].shape[0]
)
```

```python
# Remove records without a usable date or groundwater level
groundwater_clean = groundwater_df.dropna(
    subset=["Date", "WL_mbgl"]
).copy()

# Remove exact duplicate records
groundwater_clean = groundwater_clean.drop_duplicates().reset_index(drop=True)

print("Original rows:", len(groundwater_df))
print("Cleaned rows:", len(groundwater_clean))
print("Rows removed:", len(groundwater_df) - len(groundwater_clean))

print("\nRemaining missing values:")
print(groundwater_clean.isna().sum())

print("\nDate range:")
print(
    groundwater_clean["Date"].min(),
    "to",
    groundwater_clean["Date"].max()
)
```

```python
print("Records by year:")
print(
    groundwater_clean["Date"]
    .dt.year
    .value_counts()
    .sort_index()
)
```

```python
year_counts = (
    groundwater_clean["Date"]
    .dt.year
    .value_counts()
    .sort_index()
)

print(year_counts.to_string())


print("Total years:", len(year_counts))
print("Years:", year_counts.index.tolist())
```

```python
print("Groundwater records by year:")
print(year_counts.to_string())

print("\nRecords per year:")
print("Minimum:", year_counts.min())
print("Maximum:", year_counts.max())
print("Average:", round(year_counts.mean(), 2))
print("Median:", year_counts.median())
```

```python
tn_rainfall = rainfall_ds.sel(
    LATITUDE=slice(8.0, 13.5),
    LONGITUDE=slice(76.0, 80.5)
)

print(tn_rainfall)
```

```python

rain = tn_rainfall["RAINFALL"]

print("Rainfall shape:", rain.shape)
print("Minimum:", float(rain.min().values))
print("Maximum:", float(rain.max().values))
print("Mean:", float(rain.mean().values))
print("Missing values:", int(rain.isnull().sum().values))
```

```python
missing_by_location = rain.isnull().sum(dim="TIME")

print(missing_by_location)
```

```python
print("Total spatial cells:", missing_by_location.size)

print("Cells with complete rainfall data:",
      int((missing_by_location == 0).sum().values))

print("Cells with no rainfall data:",
      int((missing_by_location == 366).sum().values))

print("Cells with partially missing data:",
      int(((missing_by_location > 0) & (missing_by_location < 366)).sum().values))
```
